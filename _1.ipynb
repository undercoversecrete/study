```python
spisoc = [1,2,3,4,5]

for i in range(0, len(spisoc)- 1,2):
  spisoc[i], spisoc[i+1] = spisoc[i+1],spisoc[i]

print(spisoc)
```

```python
spisoc = [1,2,3,4,5]
n = 1

right =  spisoc[-n:] + spisoc[:-n]

print(right)
```

```python
spisoc = [1,2,2,3,3,3]

spisoc_min = min(spisoc,key=spisoc.count)

print(spisoc_min)
```

```python
spisoc = [1,2,2,3,3,3]

spisoc_max = max(spisoc,key=spisoc.count)

print(spisoc_max)
```

```python
spisok = [5,3,1,2,5,4]

median = min(spisok, key=lambda x: sum(abs(x - y) for y in spisok))

print(median)
```
